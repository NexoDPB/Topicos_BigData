### Loops

```python
palabras = cadena_1.strip(',').split()
palabras
```

```python
for palabra in palabras:
    print(palabra)
```

```python
# colocar numeracion a cada palabra
for idx, palabra in enumerate(palabras):
    print(idx, palabra)
```
